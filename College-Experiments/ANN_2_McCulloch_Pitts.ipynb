# **ANN 2 – Generate AND NOT function using McCulloch-Pitts neural net by a python program**

```python
import numpy as np
import pandas as pd
from tabulate import tabulate
```

```python
class McCullochPittsNeuron:
    """
    A simple model of a McCulloch-Pitts neuron.
    """
    def __init__(self, weights, threshold):
        """
        Initializes the neuron.
        Args:
            weights (list or np.array): The weights for the input connections.
            threshold (float): The activation threshold.
        """
        self.weights = np.array(weights)
        self.threshold = threshold

    def fire(self, inputs):
        """
        Calculates the output of the neuron.
        Args:
            inputs (list or np.array): The input signals.
        Returns:
            int: 1 if the neuron fires, 0 otherwise.
        """
        weighted_sum = np.dot(self.weights, inputs)
        return 1 if weighted_sum >= self.threshold else 0
```

```python
# --- ANDNOT Function Implementation ---

# To implement A AND (NOT B), we need the neuron to fire only when A=1 and B=0.
# Weights: w_A = 1, w_B = -1
# Threshold: theta = 1
andnot_weights = [1, -1]
andnot_threshold = 1
```

```python
# Create the neuron for the ANDNOT function
andnot_neuron = McCullochPittsNeuron(weights=andnot_weights, threshold=andnot_threshold)
```

```python
# --- Verification ---

# Generate a truth table to verify the function
test_inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]
results = []

for inputs in test_inputs:
    output = andnot_neuron.fire(inputs)
    results.append({
        'A': inputs[0],
        'B': inputs[1],
        'A AND (NOT B)': output
    })
```

```python
# Display the truth table using pandas for clear formatting
truth_table = pd.DataFrame(results)
print("--- Truth Table for ANDNOT Neuron ---")
print(truth_table.to_markdown(index=False))
```
